```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle,Arc
cir = Circle((3,3),2,color='#ffde34')
smile = Arc((3,3),2.8,2,theta1=220,theta2=-40)
eye1 = Arc((2.4,3.5),0.4,1)
eye2 = Arc((3.6,3.5),0.4,1)

plt.gca().add_patch(cir)
plt.gca().add_patch(eye1)
plt.gca().add_patch(eye2)
plt.gca().add_patch(smile)
plt.axis("equal")
plt.show()
```
