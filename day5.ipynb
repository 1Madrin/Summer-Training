```python
import numpy as np

a = np.array([1,2,3])
b = np.array([2,4,5])

print(a+b)
print(a-b)
print(a/b)
print(a//b)
print(a*b)
print(a**b)
print(a%b)
```

```python
a = np.array([[1,2,3],[2,3,4]])
b = np.array([[2,3,4],[3,4,6]])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
np.dot(a,np.transpose(b))
```

```python
a = np.array([[10,20,30],[5,10,15]])

print(np.min(a))
print(np.sum(a))
print(np.max(a))
print(np.sum(a, axis=0))
print(np.sum(a, axis=1))
```
