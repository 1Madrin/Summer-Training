```python
#pythagorean triplets
a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))
c = int(input("Enter the third number: "))

if((a*a)+(b*b)==(c*c)):
    print("the given numbers are pythagorian triplets")
```

```python
#prime numbers between 1 and 100
for num in range(1, 101):
    
    factor_count = 0
    
    for i in range(1, num + 1):
        if num % i == 0:
            factor_count += 1
            
    if factor_count == 2:
        print(num,end=" ")
    
```

```python
P = 1000
R = 10
T = 3
def SI():
    return (P*R*T)/100


def CI():
    ci = P*(pow((1+R/100),T))-P
    return ci

print(SI())
print(f"{CI():.2f}")
print(f"difference between SI and CI is: {(CI()-SI()):.2f}")
```

```python
r = float(input("Enter the radius of circle: "))
def areaCircle():
    return 3.14 *r*r

def parametereCircle():
    return 2*3.14*r

print("Aread:", areaCircle())
print(f"Parameter: {parametereCircle():.2f}" )
```
