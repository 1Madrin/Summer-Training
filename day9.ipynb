```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle



x1 = [2,2,8,8,2]
y1 = [2,6,6,2,2]
x2 = [2,3,4]
y2 = [6,8,6]
x3 = [3,7,8]
y3 = [8,8,6]
x4 = [4,4]
y4 = [2,6]
x5 = [2.5,2.5,3.5,3.5]
y5 = [2,4,4,2]
x6 = [5,5,7,7,5]
y6 = [3,5,5,3,3]
x7 = [6,6]
y7 = [3,5]
x8 = [5,7]
y8 = [4,4]


Cir = Circle([3,6.75],0.4,fill=False)
plt.gca().add_patch(Cir)

plt.axis("equal")
plt.plot(x1,y1)
plt.plot(x2,y2)
plt.plot(x3,y3)
plt.plot(x4,y4)
plt.plot(x5,y5)
plt.plot(x6,y6)
plt.plot(x7,y7)
plt.plot(x8,y8)
plt.show()
```
