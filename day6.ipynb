```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("image.jpeg")

img = np.array(img)
plt.imshow(img)
plt.axis("off")
plt.show()


X = int(img.shape[1]/2)
Y = int(img.shape[0]/2)
# img[:,:,:] = 255
# img[0:Y,X:,0] = np.clip(img[0:Y,X:,0] + 200,0,255)
img[0:Y,X:,0] = 0 
# img[0:Y,0:X,1] = np.clip(img[0:Y,0:X,1] +200)
img[0:Y,0:X,1] = 0
# img[Y:,0:X,2] = np.clip(img[Y:,0:X,2]+200)
img[Y:,0:X,2] = 0 
img[Y:,X:,:] = 0 

plt.imshow(img)
plt.axis("off")
plt.show()
```
